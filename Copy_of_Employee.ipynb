```python
df = df.drop(columns=['EmployeeCount', 'StandardHours', 'Over18','EmployeeNumber'])
df.shape
```

```python
df.duplicated().sum()
```

```python
df[df['YearsInCurrentRole'] > df['YearsAtCompany']].shape[0]
df[df['YearsWithCurrManager'] > df['YearsAtCompany']].shape[0]
df[df['YearsSinceLastPromotion'] > df['YearsAtCompany']].shape[0]
```

```python
numeric_cols = ['Age', 'DailyRate', 'HourlyRate', 'MonthlyIncome', 'MonthlyRate',
                'DistanceFromHome', 'PercentSalaryHike', 'TotalWorkingYears',
                'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion',
                'YearsWithCurrManager', 'NumCompaniesWorked', 'TrainingTimesLastYear']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outlier_count = df[(df[col] < lower_bound) | (df[col] > upper_bound)].shape[0]
    print(col, ":", outlier_count)
```

```python
df['TrainingTimesLastYear'].value_counts().sort_index()
```

```python
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    print(col, ":", df[col].unique())
    print()
```

```python
df['Attrition'] = df['Attrition'].map({'Yes': 'Left', 'No': 'Stayed'})
df['Attrition'].value_counts()
```

```python
numeric_features = ['YearsAtCompany', 'TotalWorkingYears', 'DistanceFromHome']

for col in numeric_features:
    sns.boxplot(x='Attrition', y=col, data=df)
    plt.title(f'{col} by Attrition')
    plt.show()

    print(col, "- Median:")
    print(df.groupby('Attrition')[col].median())
    print(col, "- Mean:")
    print(df.groupby('Attrition')[col].mean())
    print()
```

```python
df['Attrition_numeric'] = df['Attrition'].map({'Stayed': 0, 'Left': 1})
df.groupby(['OverTime', 'JobSatisfaction'])['Attrition_numeric'].mean() * 100
```
